### Calculation of Orientation & Displacement for Goal Position

```python
def calculate_movement_and_turns(current_pos, current_orientation, goal_pos, goal_orientation):
    # Extract current and goal positions
    x, y = current_pos
    theta = current_orientation
    x_goal, y_goal = goal_pos
    theta_goal = goal_orientation

    # Calculate angle to the goal position
    angle_to_goal = math.degrees(math.atan2(y_goal - y, x_goal - x))

    # Calculate the initial turn angle
    initial_turn_angle = angle_to_goal - theta
    # Normalize the angle to [-180, 180] range
    initial_turn_angle = (initial_turn_angle + 180) % 360 - 180

    # Calculate the distance to the goal position
    distance_to_goal = math.sqrt((x_goal - x) ** 2 + (y_goal - y) ** 2)

    # Calculate the final turn angle to match the goal orientation
    final_turn_angle = theta_goal - angle_to_goal
    # Normalize the angle to [-180, 180] range
    final_turn_angle = (final_turn_angle + 180) % 360 - 180

    return initial_turn_angle, distance_to_goal, final_turn_angle
```

```python

# Example usage
current_pos = (1,1)
current_orientation = 45
goal_pos = (2,5)
goal_orientation = 0

initial_turn, distance, final_turn = calculate_movement_and_turns(current_pos, current_orientation, goal_pos, goal_orientation)
print(f"Initial Turn Angle: {initial_turn} degrees") #alpha
print(f"Distance to Move: {distance}")
print(f"Final Turn Angle: {final_turn} degrees")
```
